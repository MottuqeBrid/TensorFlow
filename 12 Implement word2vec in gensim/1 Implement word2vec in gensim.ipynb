```python
import gensim
import pandas as pd
```

### Reading and Exploring the Dataset

The dataset we are using here is a subset of Amazon reviews from the Cell Phones & Accessories category. The data is stored as a JSON file and can be read using pandas.

Link to the Dataset: http://snap.stanford.edu/data/amazon/productGraph/categoryFiles/reviews_Cell_Phones_and_Accessories_5.json.gz


for unzip datasets run:
gunzip reviews_Cell_Phones_and_Accessories_5.json.gz


```python
df = pd.read_json("reviews_Cell_Phones_and_Accessories_5.json", lines=True)
df.head()
```

```python
df.shape
```

```python
df.reviewText[0]
```

```python
gensim.__version__
```

```python
review_text = df.reviewText.apply(gensim.utils.simple_preprocess)
```

```python
review_text
```

```python
model = gensim.models.Word2Vec(window=10, min_count=2, workers=4)
```

```python
model.build_vocab(review_text, progress_per=1000)
```

```python
model.epochs
```

```python
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.save("./word2vec-amazon-call-accessories-reviews-short.model")
```

```python
model.wv.most_similar("bad")
```

```python
model.wv.similarity("cheap", "inexpensive")
```

```python
model.wv.similarity("great", "product")
```

```python
model.wv.similarity("great", "great")
```
