# Dropout Regularization


```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
import warnings

warnings.filterwarnings("ignore")

df = pd.read_csv("./sonar_dataset.csv", header=None)
```

```python
df.sample(5)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df[60].value_counts()
```

```python
x = df.drop(60, axis=1)
y = df[60]
```

```python
y.head()
```

```python
y = pd.get_dummies(y, drop_first=True).astype("int")
y.sample(5)
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=1
)
```

```python
x_train.shape, x_test.shape
```

```python
model = keras.Sequential(
    [
        keras.layers.Dense(60, input_dim=x_train.shape[1], activation="relu"),
        keras.layers.Dense(30, activation="relu"),
        keras.layers.Dense(15, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
model.fit(x_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(x_test, y_test)
```

```python
y_pred = model.predict(x_test).reshape(-1)
print(y_pred[:10])

# round the predictions to get binary output
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
modeld = keras.Sequential(
    [
        keras.layers.Dense(60, input_dim=x_train.shape[1], activation="relu"),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(30, activation="relu"),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(15, activation="relu"),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
modeld.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
modeld.fit(x_train, y_train, epochs=100, batch_size=8)
```

```python
modeld.evaluate(x_test, y_test)
```

```python
y_predd = modeld.predict(x_test).reshape(-1)
print(y_predd[:10])

# round the predictions to get binary output
y_predd = np.round(y_predd)
print(y_predd[:10])
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(classification_report(y_test, y_predd))
```
